```python
import pandas as pd
import unicodedata

def remover_acentos(texto):
    if pd.isna(texto):
        return texto
    return unicodedata.normalize('NFKD', str(texto)).encode('ASCII', 'ignore').decode('ASCII')

# Leitura: encoding e separador do Excel BR
df = pd.read_csv('../base_csv/original/OK_estoque.csv', encoding='latin1', sep=';')

# Remove acento de todas as colunas e deixar 
df.columns = [remover_acentos(col).lower() for col in df.columns]

# Salva já formatado pro Excel/Supabase
df.to_csv('../base_csv/tratado/estoque_tratado.csv', index=False, sep=';', encoding='utf-8-sig')

print("Tratamento concluído! Arquivo salvo em base_csv/tratado/")
```
