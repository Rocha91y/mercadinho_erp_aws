```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.utils import get_column_letter

caminho = r"C:\Users\FERNANDO\Documents\Projetos\Mercadinho_ERP\1base_mestra\Base_de_dados_mercadinho - Copia.xlsx"

colunas = ["pagamento", "hora", "cliente", "vendedor"]

df = pd.read_excel(caminho, sheet_name="VENDAS", header=1)
df.columns = df.columns.str.strip().str.lower()
novo = df.groupby("id_venda")[colunas].transform("first")

wb = load_workbook(caminho)
ws = wb["VENDAS"]

for c in colunas:
    n = list(df.columns).index(c) + 1
    print(c, get_column_letter(n), ws.cell(row=2, column=n).value)

for c in colunas:
    n = list(df.columns).index(c) + 1
    for i, valor in enumerate(novo[c]):
        ws.cell(row=i + 3, column=n, value=None if pd.isna(valor) else valor)

wb.save(caminho)
```
